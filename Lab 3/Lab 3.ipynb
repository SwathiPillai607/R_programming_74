```r
transactions <- read_csv("transactions.csv")

products <- fromJSON("products.json") %>%
  as.data.frame()

customers <- read_excel("customers.xlsx")

head(transactions)
head(products)
head(customers)
```

```r
transactions_clean <- transactions %>%
  distinct() %>%
  filter(
    !is.na(InvoiceNo),
    !is.na(StockCode),
    !is.na(CustomerID),
    !is.na(Quantity),
    !is.na(InvoiceDate),
    Quantity > 0
  )

cat("Rows before cleaning:", nrow(transactions), "\n")
cat("Rows after cleaning:", nrow(transactions_clean), "\n")
```

```r
products_clean <- products %>%
  filter(
    !is.na(StockCode),
    !is.na(Description),
    !is.na(UnitPrice),
    UnitPrice > 0
  ) %>%
  group_by(StockCode) %>%
  summarise(
    Description = first(Description),
    UnitPrice = first(UnitPrice),
    .groups = "drop"
  )
cat("Unique product records:", nrow(products_clean), "\n")
```

```r
customers_clean <- customers %>%
  filter(
    !is.na(CustomerID),
    !is.na(Country)
  ) %>%
  group_by(CustomerID) %>%
  summarise(
    Country = first(Country),
    .groups = "drop"
  )

cat("Rows before cleaning:", nrow(customers), "\n")
cat("Rows after cleaning:", nrow(customers_clean), "\n")
```

```r
transactions_clean <- transactions_clean %>%
  mutate(
    StockCode = as.character(StockCode),
    CustomerID = as.character(CustomerID),
    Quantity = as.numeric(Quantity),
    InvoiceDate = as.character(InvoiceDate)
  )
products_clean <- products_clean %>%
  mutate(
    StockCode = as.character(StockCode),
    UnitPrice = as.numeric(UnitPrice)
  )
customers_clean <- customers_clean %>%
  mutate(
    CustomerID = as.character(CustomerID)
  )
```

```r
transactions_products <- transactions_clean %>%
  left_join(
    products_clean,
    by = "StockCode"
  ) %>%
  mutate(
    Revenue = Quantity * UnitPrice
  )
```

```r
summary(transactions_clean)
summary(products_clean)
summary(customers_clean)
cat("\nFinal transaction rows:", nrow(transactions_clean))
cat("\nFinal product rows:", nrow(products_clean))
cat("\nFinal customer rows:", nrow(customers_clean))
```

### Cleaning Decisions

1. Duplicate transaction records were removed.
2. Transactions with missing essential fields were removed.
3. Transactions with quantity less than or equal to zero were removed.
4. Products with missing StockCode, Description or UnitPrice were removed.
5. Products with UnitPrice less than or equal to zero were removed.
6. Product records were reduced to one record per StockCode for reliable joining.
7. Customers with missing CustomerID or Country were removed.
8. Customer records were reduced to one record per CustomerID for reliable joining.
9. Revenue was calculated as Quantity × UnitPrice.

```r
final_data <- transactions_products %>%
  left_join(
    customers_clean,
    by = "CustomerID"
  )

cat("Transactions:", nrow(transactions_clean), "\n")
cat("Final integrated dataset:", nrow(final_data), "\n")
```

```r
total_revenue <- final_data %>%
  summarise(
    Total_Revenue = sum(Revenue, na.rm = TRUE)
  )

total_revenue
```

```r
top_5_products <- final_data %>%
  filter(!is.na(Description)) %>%
  group_by(StockCode, Description) %>%
  summarise(
    Total_Revenue = sum(Revenue, na.rm = TRUE),
    .groups = "drop"
  ) %>%
  arrange(desc(Total_Revenue)) %>%
  slice_head(n = 5)

top_5_products
```

```r
top_5_countries <- final_data %>%
  filter(!is.na(Country)) %>%
  group_by(Country) %>%
  summarise(
    Total_Revenue = sum(Revenue, na.rm = TRUE),
    .groups = "drop"
  ) %>%
  arrange(desc(Total_Revenue)) %>%
  slice_head(n = 5)

top_5_countries
```

```r
top_5_customers <- final_data %>%
  filter(!is.na(CustomerID)) %>%
  group_by(CustomerID) %>%
  summarise(
    Total_Purchase_Value = sum(Revenue, na.rm = TRUE),
    .groups = "drop"
  ) %>%
  arrange(desc(Total_Purchase_Value)) %>%
  slice_head(n = 5)

top_5_customers
```

```r
customer_value <- final_data %>%
  filter(!is.na(CustomerID)) %>%
  group_by(CustomerID) %>%
  summarise(
    Total_Purchase_Value = sum(Revenue, na.rm = TRUE),
    .groups = "drop"
  )
head(customer_value)
```

```r
q1 <- quantile(
  customer_value$Total_Purchase_Value,
  0.25,
  na.rm = TRUE
)
q2 <- quantile(
  customer_value$Total_Purchase_Value,
  0.50,
  na.rm = TRUE
)
q3 <- quantile(
  customer_value$Total_Purchase_Value,
  0.75,
  na.rm = TRUE
)
cat("Low/Medium threshold:", q1, "\n")
cat("Medium/High threshold:", q2, "\n")
cat("High/Premium threshold:", q3, "\n")
```

```r
customer_value <- customer_value %>%
  mutate(
    Customer_Segment = case_when(
      Total_Purchase_Value <= q1 ~ "Low Value",
      Total_Purchase_Value <= q2 ~ "Medium Value",
      Total_Purchase_Value <= q3 ~ "High Value",
      TRUE ~ "Premium"
    )
  )
head(customer_value)
```

```r
customer_segments <- customer_value %>%
  count(Customer_Segment)
customer_segments
```

```r
segment_revenue <- customer_value %>%
  group_by(Customer_Segment) %>%
  summarise(
    Number_of_Customers = n(),
    Total_Revenue = sum(Total_Purchase_Value),
    Average_Purchase_Value = mean(Total_Purchase_Value),
    .groups = "drop"
  ) %>%
  arrange(desc(Total_Revenue))
segment_revenue
```

```r
country_performance <- final_data %>%
  filter(!is.na(Country)) %>%
  group_by(Country) %>%
  summarise(
    Total_Revenue = sum(Revenue, na.rm = TRUE),
    .groups = "drop"
  ) %>%
  arrange(desc(Total_Revenue))

country_performance
```

```r
high_performing_market <- country_performance %>%
  slice_max(
    order_by = Total_Revenue,
    n = 1
  )
high_performing_market
```

```r
underperforming_market <- country_performance %>%
  slice_min(
    order_by = Total_Revenue,
    n = 1
  )
underperforming_market
```

```r
query2 <- "
SELECT
    Country,
    SUM(Revenue) AS Total_Revenue
FROM retail_sales
WHERE Country IS NOT NULL
GROUP BY Country
ORDER BY Total_Revenue DESC;
"

revenue_by_country_sql <- dbGetQuery(con, query2)

head(revenue_by_country_sql, 10)
```

## Three Business Insights

### 1. United Kingdom is the strongest market
The **United Kingdom** is the highest-performing market, generating total revenue of **8,977,081**. This represents the largest share of the overall revenue of **10,885,988**, indicating that the United Kingdom is the company's primary market.

### 2. Premium customers contribute the majority of revenue
The **Premium** customer segment generated **8,630,925** in revenue from **1,085 customers**, with an average purchase value of approximately **7,955** per customer. This indicates that premium customers make a major contribution to overall revenue and should be prioritized for customer retention and loyalty strategies.

### 3. Certain products and customers contribute significantly to sales
The top-performing product, **PAPER CRAFT , LITTLE BIRDIE (StockCode: 23843)**, generated **168,470** in revenue. The highest-value customer, **Customer 18102**, generated **408,968** in purchases. This shows that specific products and high-value customers contribute significantly to sales and should receive attention in inventory planning, promotions, and customer retention.
